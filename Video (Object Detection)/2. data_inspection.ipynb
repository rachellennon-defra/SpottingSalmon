```python
import os
import cv2
import matplotlib.pyplot as plt
import random
```

#### Step 1: Check files and images are as expected

```python
# Check train, valid and test files have expected number of files
base_dir = "/dbfs/mnt/lab/unrestricted/rachel.lennon@defra.gov.uk/videos/LabelledFishYOLOv11" # Path to labelled images from data labelling stage.
for split in ["train", "valid", "test"]:
    split_dir = os.path.join(base_dir, split)
    if os.path.exists(split_dir):
        print(f"{split} has {len(os.listdir(split_dir))} files")
    else:
        print(f"{split} directory not found!")
```

```python
# Pick 3 random images from train to inspect
train_dir = os.path.join(base_dir, "/dbfs/mnt/lab/unrestricted/rachel.lennon@defra.gov.uk/videos/LabelledFishYOLOv11/train/images")
files = [f for f in os.listdir(train_dir) if f.endswith((".jpg", ".png"))]

print("Total training images:", len(files))

for i in random.sample(files, min(3, len(files))):
    img_path = os.path.join(train_dir, i)
    img = cv2.imread(img_path)
    if img is None:
        print(f"⚠️ Could not read {i}")
        continue

    h, w, c = img.shape
    print(f"{i}: {w}x{h}, channels={c}")

    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title(i)
    plt.axis("off")
    plt.show()
```

```python
# Inspect labels to ensure they are in correct YOLO format

label_dir = os.path.join(base_dir, "/dbfs/mnt/lab/unrestricted/rachel.lennon@defra.gov.uk/videos/LabelledFishYOLOv11/train/labels")
label_files = os.listdir(label_dir)[:5]

# class id, x_center, y_center, width, height
for lf in label_files:
    print(f"File: {lf}")
    with open(os.path.join(label_dir, lf)) as f:
        lines = f.readlines()
        print("  ", lines[:3])  
```
